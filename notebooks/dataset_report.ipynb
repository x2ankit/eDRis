# Dataset Audit Report
This notebook analyzes the datasets available for eDRis as per Phase 1 of the implementation plan.


## 1. APTOS 2019 Dataset
- Total Images
- Class Distribution (Level 0-4)
- Image Dimensions
- Image Quality


```
import pandas as pd
import os
import glob
from PIL import Image

aptos_train_csv = 'datasets/classification/aptos2019/train.csv'
if os.path.exists(aptos_train_csv):
    aptos_train_df = pd.read_csv(aptos_train_csv)
    print(f"APTOS Total Training Images: {len(aptos_train_df)}")
    print("Class Distribution:")
    print(aptos_train_df['diagnosis'].value_counts().sort_index())
else:
    print("APTOS train.csv not found.")

aptos_images = glob.glob('datasets/classification/aptos2019/**/*.png', recursive=True)
print(f"Total APTOS .png images found: {len(aptos_images)}")
if aptos_images:
    with Image.open(aptos_images[0]) as img:
        print(f"Sample Image Dimensions: {img.size}")

```
