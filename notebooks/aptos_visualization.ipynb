# APTOS Dataset Visualization
This notebook visualizes the APTOS 2019 dataset, class distributions, and sample fundus images across different DR severity grades.


```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from PIL import Image

train_csv_path = 'datasets/classification/aptos2019/train.csv'
train_dir = 'datasets/classification/aptos2019/train_images'

df = pd.read_csv(train_csv_path)
print(f"Total labeled images: {len(df)}")

```

## Class Distribution


```
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='diagnosis', palette='viridis')
plt.title('APTOS 2019 DR Severity Distribution')
plt.xlabel('DR Grade (0: No DR, 1: Mild, 2: Moderate, 3: Severe, 4: Proliferative)')
plt.ylabel('Count')
plt.show()

```

## Sample Images per Grade


```
fig, axes = plt.subplots(1, 5, figsize=(20, 5))
for grade in range(5):
    sample = df[df['diagnosis'] == grade].sample(1).iloc[0]
    img_path = os.path.join(train_dir, f"{sample['id_code']}.png")
    
    if os.path.exists(img_path):
        img = Image.open(img_path)
        axes[grade].imshow(img)
        axes[grade].set_title(f"Grade {grade}")
        axes[grade].axis('off')
    else:
        axes[grade].set_title(f"Image Missing\nGrade {grade}")
        axes[grade].axis('off')

plt.tight_layout()
plt.show()

```
